```python
def optimal_page_replacement(pages, capacity):
    memory = []
    page_faults = 0

    for i in range(len(pages)):
        if pages[i] not in memory:
            page_faults += 1
            if len(memory) == capacity:
                # Find the page that will not be used for the longest time in the future
                farthest = 0
                index_to_remove = 0
                for j in range(len(memory)):
                    if memory[j] not in pages[i + 1:]:
                        index_to_remove = j
                        break
                    else:
                        next_use = pages[i + 1:].index(memory[j])
                        if next_use > farthest:
                            farthest = next_use
                            index_to_remove = j
                memory.pop(index_to_remove)  # Remove the optimal page
            memory.append(pages[i])  # Add the new page
    return page_faults

# Example Usage
pages = [7, 0, 1, 2, 0, 3, 0, 4, 2, 3, 0, 3, 2]
capacity = 3
print("Optimal Page Faults:", optimal_page_replacement(pages, capacity))
```
